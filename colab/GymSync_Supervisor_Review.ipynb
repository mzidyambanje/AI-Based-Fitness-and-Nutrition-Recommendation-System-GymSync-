## 2. Project Structure Overview

````python
import os
from pathlib import Path

def show_tree(path, prefix="", max_depth=3, current_depth=0, ignore_dirs={"build", ".git", ".dart_tool", "node_modules", ".__pycache__"}):
    if current_depth >= max_depth:
        return
    
    try:
        items = sorted(os.listdir(path))
        items = [i for i in items if i not in ignore_dirs and not i.startswith('.')]
        
        for i, item in enumerate(items):
            item_path = os.path.join(path, item)
            is_last = i == len(items) - 1
            current_prefix = "└── " if is_last else "├── "
            print(f"{prefix}{current_prefix}{item}")
            
            if os.path.isdir(item_path) and item not in ignore_dirs:
                next_prefix = prefix + ("    " if is_last else "│   ")
                show_tree(item_path, next_prefix, max_depth, current_depth + 1, ignore_dirs)
    except PermissionError:
        pass

print("\n📁 GymSync Project Structure:")
print("="*50)
show_tree(".")
````

## 3. Research Methodology: DDR Phases

````python
# Display the methodology document
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch
import numpy as np

phases = [
    {"num": 1, "name": "Literature Review", "output": "Research Gap Identified", "status": "✓"},
    {"num": 2, "name": "Requirements Analysis", "output": "System Requirements Defined", "status": "✓"},
    {"num": 3, "name": "Data Collection", "output": "User Attribute Dataset", "status": "✓"},
    {"num": 4, "name": "System Design", "output": "System Architecture", "status": "⏳"},
    {"num": 5, "name": "System Development", "output": "Flutter-Firebase Prototype", "status": "✓"},
    {"num": 6, "name": "Recommendation Engine", "output": "Hybrid Recommendation System", "status": "⏳"},
    {"num": 7, "name": "Testing & Evaluation", "output": "Results and Validation", "status": "⏳"},
]

fig, ax = plt.subplots(figsize=(14, 8))
ax.set_xlim(0, 14)
ax.set_ylim(0, 9)
ax.axis('off')

# Title
ax.text(7, 8.5, "Design and Development Research (DDR) Approach", 
        ha="center", fontsize=16, fontweight="bold")

# Draw phases
for i, phase in enumerate(phases):
    x = (i % 4) * 3.3 + 0.5
    y = 7 - (i // 4) * 3.5
    
    # Color based on status
    if phase["status"] == "✓":
        color = "#90EE90"  # Light green
    else:
        color = "#FFE4B5"  # Light orange
    
    # Phase box
    box = FancyBboxPatch((x, y-1.2), 2.8, 2.2, 
                         boxstyle="round,pad=0.1", 
                         edgecolor="black", facecolor=color, linewidth=2)
    ax.add_patch(box)
    
    # Phase number and name
    ax.text(x+1.4, y+0.7, f"Phase {phase['num']}", 
           ha="center", fontsize=11, fontweight="bold")
    ax.text(x+1.4, y+0.3, phase['name'], 
           ha="center", fontsize=9, style="italic", wrap=True)
    
    # Output
    ax.text(x+1.4, y-0.4, phase['output'], 
           ha="center", fontsize=7, style="italic", color="#333333")
    
    # Status
    ax.text(x+2.6, y+0.9, phase['status'], 
           ha="right", fontsize=12, fontweight="bold")

# Legend
ax.text(0.5, 0.5, "✓ = Completed", fontsize=10, bbox=dict(boxstyle="round", facecolor="#90EE90"))
ax.text(3, 0.5, "⏳ = In Progress/Planned", fontsize=10, bbox=dict(boxstyle="round", facecolor="#FFE4B5"))

plt.tight_layout()
plt.savefig('methodology_phases.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n📊 Research Methodology Status:")
print("="*60)
for phase in phases:
    print(f"Phase {phase['num']}: {phase['name']:<25} {phase['status']}")
````
