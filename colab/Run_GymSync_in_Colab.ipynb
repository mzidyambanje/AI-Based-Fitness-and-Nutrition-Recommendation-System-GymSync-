```python
# Display the first 400 lines of `lib/main.dart` for quick inspection
import pathlib
p = pathlib.Path('AI-Based-Fitness-and-Nutrition-Recommendation-System-GymSync-') / 'lib' / 'main.dart'
if p.exists():
    print('Showing:', p)
    with p.open('r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i >= 400:
                break
            print(line.rstrip())
else:
    print('File not found:', p)
```

```python
# If the repo is private, clone using a GitHub Personal Access Token (PAT) securely
# This cell will prompt for a token and clone using subprocess (safer than embedding token in URL).
import getpass, subprocess, shlex
token = getpass.getpass('GitHub PAT (leave empty for public repos): ')
if token:
    repo = 'https://github.com/mzidyambanje/AI-Based-Fitness-and-Nutrition-Recommendation-System-GymSync-.git'
    url = f'https://{token}@' + repo.split('https://',1)[1]
    print('Cloning private repo...')
    subprocess.run(shlex.split(f'git clone {url}'))
else:
    print('No token provided — skip private clone.')
```
